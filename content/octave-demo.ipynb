# Octave in the browser

This runs real GNU Octave, compiled to WebAssembly, on your own machine.
Nothing is uploaded. The first run downloads Octave, which takes a moment;
after that it is cached.

Use **Save .m** on the toolbar to download this as a script MATLAB will open,
and **Open .m** to bring your own script in.

```octave
A = [1 2; 3 4];
disp(A * A)
```

```octave
x = 1:10;
plot(x, x.^2);
xlabel('x');
ylabel('x squared');
```

## Octave syntax MATLAB will not take

The cell below is valid Octave and invalid MATLAB. Press **Save .m** and the
downloaded file has it converted, with a note saying what changed.

```octave
total = 0;
for i = 1:5
  if mod(i, 2) != 0
    total += i;
  endif
endfor
printf('%d\n', total);
```
